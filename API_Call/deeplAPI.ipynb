```python
import requests



BASE = "https://api-free.deepl.com/v2/translate"
```

```python
import dotenv
import os

dotenv.load_dotenv()
DEEPL_API_KEY = os.getenv("DEEPL_API")
print(DEEPL_API_KEY)
```

```python
API_KEY = DEEPL_API_KEY


prompt = "Hallo Welt, wie geht es dir? Ich hoffe, du hast einen schönen Tag!"

json={
        "text": [prompt],  # Liste, mehrere Texte pro Request möglich
        "source_lang": "DE",                      # optional, sonst automatische Erkennung
        "target_lang": "EN-US",
    }


r = requests.post(BASE, headers={"Authorization": f"DeepL-Auth-Key {API_KEY}"}, timeout=100, json= json)

print(r.status_code)
data = r.json()


print(data["translations"][0]["text"])
```

```python
import os
import requests

# Free-Keys enden auf ":fx" und nutzen api-free, Pro-Keys api.deepl.com
BASE_URL = "https://api-free.deepl.com" if API_KEY.endswith(":fx") else "https://api.deepl.com"

response = requests.post(
    f"{BASE_URL}/v2/translate",
    headers={"Authorization": f"DeepL-Auth-Key {API_KEY}"},
    json={
        "text": ["Hallo Welt, wie geht es dir?"],  # Liste, mehrere Texte pro Request möglich
        "source_lang": "DE",                      # optional, sonst automatische Erkennung
        "target_lang": "EN-US",
    },
    timeout=10,
)
print(response.status_code)



for t in response.json()["translations"]:
    print(t["text"], f"(erkannt: {t['detected_source_language']})")
```
