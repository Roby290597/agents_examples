```python
import requests


BASE = "https://rest.arbeitsagentur.de/jobboerse/jobsuche-service/pc/v6/jobs"
HEADERS = {"X-API-Key": "jobboerse-jobsuche"}


job_title = "Data Scientist"

params = {"was": job_title, "wo": "Deutschland", "umkreis": 50,
          "angebotsart": 1, "page": [1,2], "size": 30}
r = requests.get(BASE, headers=HEADERS, params=params, timeout=20)
print(r.status_code)
data = r.json()

print(data["page"])
#print(data["ergebnisliste"][0].keys())  # Struktur von v6 erst ansehen
```

```python
important_keys = ['hauptberuf', 'firma',  'stellenlokationen', 'externeURL']

for i, entry in enumerate(data["ergebnisliste"]):
    job_info = {}
    for key in important_keys:
        if key in entry.keys():
            print(key)
            if key != "stellenlokationen" and key in entry.keys():
                job_info[key] = entry[key]
            else:
                if "ort" in entry["stellenlokationen"][0]["adresse"].keys():
                    job_info["ort"] = entry["stellenlokationen"][0]["adresse"]["ort"]
                else:
                    job_info["ort"] = "N/A"
                if "region" in entry["stellenlokationen"][0]["adresse"].keys():
                    job_info["region"] = entry["stellenlokationen"][0]["adresse"]["region"]
                else:
                    job_info["region"] = "N/A"
        else:
            job_info[key] = "N/A"
    print(f"{i}: {job_info}\n")
```

```python
fields = ["hauptberuf", "firma", "externeURL", 'gehaltsspanneVon', 'gehaltsspanneBis']
jobs = []

for entry in data["ergebnisliste"]:
    job_info = {key: entry.get(key) for key in fields}

    lokation = (entry.get("stellenlokationen") or [{}])[0]
    adresse = lokation.get("adresse", {})
    job_info["ort"] = adresse.get("ort")
    job_info["region"] = adresse.get("region")

    jobs.append(job_info)
    print(f"Job Info: {job_info}\n")
```
